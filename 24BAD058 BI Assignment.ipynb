```python
print("24BAD058")
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("SuperStoreOrders.csv")
```

```python
df.head()
df.shape
df.info()
```

```python
print("Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nMissing values:")
print(df.isnull().sum())
```

```python
messy_df = df.copy()
```

```python
messy_df.loc[5, "sales"] = np.nan
messy_df.loc[20, "region"] = np.nan
messy_df.loc[50, "quantity"] = np.nan
```

```python
messy_df.isnull().sum()
```

```python
messy_df.loc[100, "sales"] = 10000000
messy_df.loc[100, ["order_id", "product_name", "sales"]]
```

```python
messy_df.loc[10, "order_date"] = "15/01/2012"
messy_df.loc[11, "order_date"] = "2012-02-20"
messy_df.loc[12, "order_date"] = "March 25, 2012"
messy_df.loc[10:12, "order_date"]
```

```python
# --- Handling Missing Values in messy_df ---

print("Original missing values:")
print(messy_df.isnull().sum())
print("\n" + "="*50 + "\n")

# Data Cleaning: Convert 'sales' and 'quantity' to numeric, handling potential non-numeric entries
# 'sales' often comes with commas as thousand separators in raw data
# 'quantity' could also be an object type if there were any non-numeric entries

messy_df['sales'] = messy_df['sales'].astype(str).str.replace(',', '', regex=False)
messy_df['sales'] = pd.to_numeric(messy_df['sales'], errors='coerce')

# Ensure 'quantity' is numeric as well, handling potential non-numeric entries
messy_df['quantity'] = pd.to_numeric(messy_df['quantity'], errors='coerce')

# 1. Handling 'sales' column (Numerical)
# Method: Impute with the median
# Reason: Sales is a numerical column, and using the median is a robust strategy
#         against potential outliers, which could skew the mean. It preserves
#         more of the original data distribution compared to dropping rows.
median_sales = messy_df['sales'].median()
messy_df['sales'] = messy_df['sales'].fillna(median_sales)
print(f"Filled missing 'sales' values with median: {median_sales:.2f}")

# 2. Handling 'quantity' column (Numerical)
# Method: Impute with the median
# Reason: Quantity is also numerical. Similar to sales, median imputation
#         is robust to outliers. Since quantity represents counts, median is
#         often a good central tendency measure. One could also consider
#         the mode, but median is a general robust choice for numerical data.
median_quantity = messy_df['quantity'].median()
messy_df['quantity'] = messy_df['quantity'].fillna(median_quantity)
print(f"Filled missing 'quantity' values with median: {median_quantity:.0f}")

# 3. Handling 'region' column (Categorical)
# Method: Impute with the mode
# Reason: Region is a categorical column. Replacing missing values with the
#         most frequent category (mode) is a common and suitable approach
#         as it maintains the distribution of the existing categories and
#         introduces the most representative value for the missing entries.
#         We use `.mode()[0]` because `.mode()` can return multiple values if there's a tie.
mode_region = messy_df['region'].mode()[0]
messy_df['region'] = messy_df['region'].fillna(mode_region)
print(f"Filled missing 'region' values with mode: {mode_region}")

print("\n" + "="*50 + "\n")
print("Missing values after imputation:")
print(messy_df.isnull().sum())
```

```python
Q1 = messy_df["sales"].quantile(0.25)
Q3 = messy_df["sales"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = messy_df[
    (messy_df["sales"] < lower_bound) |
    (messy_df["sales"] > upper_bound)
]

print(outliers[["order_id", "sales"]])
```

```python
print(messy_df.isnull().sum())
```

```python
messy_df["sales"] = messy_df["sales"].clip(
    lower=lower_bound,
    upper=upper_bound
)
print(messy_df["sales"].max())
```

```python
messy_df["order_date"] = pd.to_datetime(
    messy_df["order_date"],
    errors="coerce"
)
print(messy_df["order_date"].head())
```

```python
print(
    "Invalid dates:",
    messy_df["order_date"].isnull().sum()
)
```

```python
messy_df["year"] = messy_df["order_date"].dt.year

messy_df["quarter"] = (
    "Q" + messy_df["order_date"].dt.quarter.astype(str)
)

messy_df["month"] = messy_df["order_date"].dt.month

messy_df["month_name"] = messy_df["order_date"].dt.month_name()
```

```python
messy_df[
    ["order_date", "year", "quarter", "month", "month_name"]
].head()
```

```python
bi_data = (
    messy_df
    .groupby(
        ["region", "year", "quarter", "month"],
        as_index=False
    )["sales"]
    .sum()
)
```

```python
region_sales = (
    messy_df
    .groupby("region")["sales"]
    .sum()
    .reset_index()
)
```

```python
region_sales["sales_rank"] = (
    region_sales["sales"]
    .rank(
        ascending=False,
        method="dense"
    )
)
```

```python
region_sales.sort_values("sales_rank")
monthly_sales = (
    messy_df
    .groupby(
        ["year", "month"],
        as_index=False
    )["sales"]
    .sum()
)
```

```python
monthly_sales = monthly_sales.sort_values(
    ["year", "month"]
)
monthly_sales["previous_month_sales"] = (
    monthly_sales["sales"].shift(1)
)
monthly_sales.head(15)
```

```python
monthly_sales["cumulative_sales"] = (
    monthly_sales["sales"].cumsum()
)
monthly_sales.head(15)
```
